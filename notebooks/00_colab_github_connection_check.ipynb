# Colab ↔ GitHub connection check

This notebook has one job: prove the Colab ↔ GitHub workflow works.

**How to use it:**
1. Open this notebook in Colab directly from GitHub (File → Open notebook → GitHub tab → paste this repo's URL, or use the `iambekzodboboev/traffic-sign-recognition` repo and pick this file).
2. Run the cell below (Runtime → Run all).
3. Check the output: Python version, whether PyTorch is available, and whether a GPU is attached.
4. Save a copy back to GitHub (File → Save a copy in GitHub) to confirm the push side works too.

```python
import sys, platform

print("Python:", sys.version)
print("Platform:", platform.platform())

try:
    import torch
    print("PyTorch:", torch.__version__)
    print("CUDA available:", torch.cuda.is_available())
    if torch.cuda.is_available():
        print("GPU:", torch.cuda.get_device_name(0))
    else:
        print("No GPU attached to this Colab runtime (check Runtime > Change runtime type).")
except ImportError:
    print("PyTorch is not installed in this environment.")
```

If you see Python/PyTorch info printed above with no errors, the Colab side works. Save a copy back to GitHub next to confirm the full round trip.
